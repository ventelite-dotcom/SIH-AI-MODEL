```python
import cv2
import pytesseract
import re
import json
import numpy as np
from pdf2image import convert_from_path

# --- THE FIX FOR MASSIVE PDFs ---
from PIL import Image
Image.MAX_IMAGE_PIXELS = None
# --------------------------------

def process_pdf(pdf_path):
    print(f"--- Processing: {pdf_path} ---")
    try:
        # Convert PDF to Image
        pages = convert_from_path(pdf_path, dpi=300)
        first_page = pages[0]

        # Convert to OpenCV format
        opencv_image = np.array(first_page)
        image = opencv_image[:, :, ::-1].copy()

        # Grayscale for better reading
        gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

        # Extract Text
        extracted_text = pytesseract.image_to_string(gray)

        # THE BRAIN: Dual Regex Pattern
        date_pattern = r"(?:Expiry|Valid Till|Valid Upto|Date of Expiry|Academic Year)[\s:-]*(\d{4}-\d{2}|\d{2}[-/]\d{2}[-/]\d{4})"
        match = re.search(date_pattern, extracted_text, re.IGNORECASE)

        if match:
            result = {
                "file_name": pdf_path,
                "status": "Approved",
                "document_valid": True,
                "extracted_date_or_year": match.group(1),
            }
        else:
            result = {
                "file_name": pdf_path,
                "status": "Flagged (Manual Review Required)",
                "document_valid": False,
                "extracted_date_or_year": None,
                "error_hint": "Could not find a valid date or year format."
            }

        return result

    except Exception as e:
         return {"error": str(e)}

# --- Run the test on BOTH documents ---
doc1_result = process_pdf('fire-safety-certificate-model-training.pdf')
doc2_result = process_pdf('model training approval letter.pdf')

final_dashboard_json = [doc1_result, doc2_result]
print(json.dumps(final_dashboard_json, indent=2))
```

```python
import fitz  # PyMuPDF
import torch
from transformers import pipeline

print("Downloading the lightweight AI model...")
# Using the 0.5B model prevents the RAM crash!
local_auditor = pipeline(
    "text-generation",
    model="Qwen/Qwen2.5-0.5B-Instruct",
    device=0, # Forces it onto the GPU
    torch_dtype=torch.float16
)
print("Model built and ready!")
```

```python
def audit_pdf_locally(pdf_path):
    print(f"\n--- 📄 Auditing {pdf_path} ---")
    doc = fitz.open(pdf_path)

    # We simplified the rule so the 0.5B model understands it easily
    aicte_rules = "Check if this document mentions an Academic Year (like 2023 or 2024) OR an Expiry Date."

    for page_num in range(min(1, len(doc))): # Just scanning page 1
        text = doc[page_num].get_text()

        # A much stricter, simpler prompt
        prompt = f"""
        Task: Audit the document.
        Rule: {aicte_rules}
        Document Text: {text[:800]}

        Did it pass the rule? Answer with ONE WORD ONLY: PASS or VIOLATION.
        """

        raw_result = local_auditor(prompt, max_new_tokens=15, do_sample=False)[0]['generated_text']

        # THE FIX: We ignore the model's chatty fluff and just search for the keywords
        if "PASS" in raw_result.upper() and "VIOLATION" not in raw_result.upper():
            print("AI Decision: ✅ APPROVED (Passed AICTE Checks)")
        elif "VIOLATION" in raw_result.upper():
            print("AI Decision: ❌ VIOLATION (Flagged for Review)")
        else:
            print("AI Decision: ⚠️ UNCERTAIN (Requires Human Check)")

# Test it again!
audit_pdf_locally("fire-safety-certificate-model-training.pdf")
audit_pdf_locally("model training approval letter.pdf")
```
